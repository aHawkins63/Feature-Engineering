# Different types of Cross Validations Techniques

## Load cancer dataset

```python
import pandas as pd
df=pd.read_csv('cancer_dataset .csv')
df.head()
```

## Remove unwanted columns

```python
df.drop(columns=['Unnamed: 32'], axis=1, inplace=True)
```

```python
df.isnull().sum().sum()
```

## split data into Independent And dependent features

```python
X=df.iloc[:,2:]
y=df.iloc[:,1]
```

### HoldOut Validation Approach- Train And Test Split

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
import numpy as np # Ensure numpy is imported

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=4)
model = DecisionTreeClassifier()
model.fit(X_train, y_train)
result = model.score(X_test, y_test)

print(f"Hold Out Cross Validation's result = {result}")
```

### K Fold Cross Validation

```python
from sklearn.model_selection import KFold
from sklearn.model_selection import cross_val_score
from sklearn.tree import DecisionTreeClassifier # Ensure model is imported
import numpy as np

model=DecisionTreeClassifier()
kfold_validation=KFold(10)
results=cross_val_score(model,X,y,cv=kfold_validation)

print(f"K-Fold Cross Validation's result = {np.mean(results)}")
```

### Stratified K-fold Cross Validation

```python
from sklearn.model_selection import StratifiedKFold
from sklearn.tree import DecisionTreeClassifier # Ensure model is imported
from sklearn.model_selection import cross_val_score # Ensure cross_val_score is imported
import numpy as np # Ensure numpy is imported

skfold=StratifiedKFold(n_splits=5)
model=DecisionTreeClassifier()
scores=cross_val_score(model,X,y,cv=skfold)

print(f"Stratified K-fold Cross Validation's result = {np.mean(scores)}")
```

### Leave One Out Cross Validation(LOOCV)

```python
from sklearn.model_selection import LeaveOneOut
from sklearn.tree import DecisionTreeClassifier # Ensure model is imported
from sklearn.model_selection import cross_val_score # Ensure cross_val_score is imported
import numpy as np # Ensure numpy is imported

model=DecisionTreeClassifier()
leave_validation=LeaveOneOut()
results=cross_val_score(model,X,y,cv=leave_validation)

print(f"Leave One Out Cross Validation's result = {np.mean(results)}")
```

### K-Fold Cross Validation for SVM and Naive Bayes

```python
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import cross_val_score
import numpy as np

# SVM with 10-Fold Cross Validation
model_svc = SVC()
results_svc_kfold = cross_val_score(model_svc, X, y, cv=kfold_validation)
print(f"SVM with K-Fold Cross Validation's result = {np.mean(results_svc_kfold)}")

# Naive Bayes with 10-Fold Cross Validation
model_nb = GaussianNB()
results_nb_kfold = cross_val_score(model_nb, X, y, cv=kfold_validation)
print(f"Naive Bayes with K-Fold Cross Validation's result = {np.mean(results_nb_kfold)}")
```
