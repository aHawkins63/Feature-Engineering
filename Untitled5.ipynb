# Animal Text Analysis: 2-Gram Predictions and Cosine Similarity

```python
import re
from collections import defaultdict
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

source_1 = """In a dense forest, a cunning fox, a fierce wolf, and a mighty lion coexisted, each with its unique strengths and strategies. The fox, known for its sly demeanor, would often cross paths with the wolf during their nightly hunts. Their interactions were a game of wits, with the fox using its clever tricks to outmaneuver the wolf, slipping through the dense underbrush like a shadow. Meanwhile, the wolf, with its keen senses and pack mentality, was a force to be reckoned with. Its haunting howls echoed through the woods, serving as both a warning and a signal to its fellow pack members. Amidst this intricate dance of predator and prey, the lion, regal and imposing, ruled over the forest like a sovereign monarch. Its mere presence commanded respect from all creatures, and the lion's deep, rumbling roar served as a reminder of its supreme authority. Together, the fox, wolf, and lion played their roles, shaping the destiny of the forest as they navigated the delicate balance between cunning, strength, and leadership."""

source_2 = """In a vast savannah, a clever fox, a gentle sheep, and a majestic lion shared a complex relationship that defied the expectations of the animal kingdom. The fox, renowned for its cunning and resourcefulness, had forged an unlikely friendship with the timid sheep. The sheep, despite being vulnerable to many predators, felt safe in the presence of the fox, for the cunning fox had vowed to protect the sheep from harm. One day, a mighty lion arrived in the savannah, claiming dominion over all its inhabitants. The lion's imposing presence struck fear into the hearts of the animals, and all the animals trembled in submission. However, the fox, guided by its cleverness, devised a plan to negotiate with the lion. Approaching the lion with the sheep by its side, the fox proposed a pact of cooperation. The fox offered the lion the wisdom of its strategies and the gentle companionship of the sheep in exchange for peace and prosperity for all. The lion, initially skeptical, recognized the potential benefits of such an arrangement. Over time, the three creatures worked together harmoniously. The fox's cunning helped the lion secure more food, and the sheep's gentle nature provided solace to the mighty lion. In return, the lion protected the fox and the sheep from external threats. Together, the fox, sheep, and lion thrived in the savannah, proving that even the most unlikely friendships could lead to prosperity, unity, and a better life for all."""
```

### Task 1: 2-Gram Word Predictions for "fox" and "lion"

```python
def clean_and_tokenize(text):
    text_clean = re.sub(r'[^a-zA-Z0-9\s]', ' ', text.lower())
    return text_clean.split()

tokens_s1 = clean_and_tokenize(source_1)
tokens_s2 = clean_and_tokenize(source_2)
all_tokens = tokens_s1 + tokens_s2

pred_words = defaultdict(list)
succ_words = defaultdict(list)

targets = {'fox', 'lion', 'lions'}

for i in range(len(all_tokens)):
    word = all_tokens[i]
    if word in targets:
        if i > 0:
            pred_words[word].append(all_tokens[i-1])
        if i < len(all_tokens) - 1:
            succ_words[word].append(all_tokens[i+1])

print("=== TASK 1: 2-GRAM PREDICTIONS ===")
for target in sorted(list(targets)):
    print(f"\nWord: '{target}'")
    print(f"  Preceded by: {dict(sorted({w: pred_words[target].count(w) for w in set(pred_words[target])}.items(), key=lambda x: -x[1]))}")
    print(f"  Succeeded by: {dict(sorted({w: succ_words[target].count(w) for w in set(succ_words[target])}.items(), key=lambda x: -x[1]))}")
```

### Task 2: Cosine Similarity between Animal Co-occurrences / Contexts

```python
sentences = re.split(r'[.!?]', source_1 + " " + source_2)
sentences = [s.strip().lower() for s in sentences if s.strip()]

animals = ['fox', 'sheep', 'wolf', 'lion']
animal_docs = {}

for animal in animals:
    mention_sentences = [s for s in sentences if re.search(rf'\b{animal}\b', s)]
    animal_docs[animal] = " ".join(mention_sentences)

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform([animal_docs[a] for a in animals])
animal_vectors = {animal: tfidf_matrix[i].toarray() for i, animal in enumerate(animals)}

def calculate_similarity(a1, a2):
    sim = cosine_similarity(animal_vectors[a1], animal_vectors[a2])[0][0]
    return sim

pairs = [
    ('fox', 'sheep'),
    ('wolf', 'lion'),
    ('wolf', 'sheep'),
    ('wolf', 'fox'),
    ('lion', 'sheep')
]

print("=== TASK 2: COSINE SIMILARITIES ===")
for p1, p2 in pairs:
    similarity = calculate_similarity(p1, p2)
    print(f"Cosine Similarity ({p1}, {p2}): {similarity:.4f}")
```
